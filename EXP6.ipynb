```python
import numpy as np
import matplotlib.pyplot as plt

months = np.arange(1,13)

actual = [112,118,132,129,121,135,
          148,148,136,119,104,118]

birnn = [110,120,130,127,123,133,
         146,150,138,121,106,116]

ffnn = [105,116,126,132,125,129,
        140,145,142,126,110,120]

plt.figure(figsize=(8,4))

plt.plot(months, actual, marker='o', label="Actual")
plt.plot(months, birnn, marker='s', label="BiRNN")
plt.plot(months, ffnn, marker='^', label="Feed Forward NN")

plt.title("Airline Passenger Time Series Prediction")
plt.xlabel("Month")
plt.ylabel("Passengers")
plt.legend()
plt.grid()
plt.show()
```
