```python
import numpy as np
import matplotlib.pyplot as plt

epochs = np.arange(1,11)

train_acc = [35,43,49,54,58,61,64,66,68,70]
test_acc  = [32,40,46,50,53,55,57,59,60,61]

plt.figure(figsize=(7,4))
plt.plot(epochs, train_acc, marker='o', label="Training Accuracy")
plt.plot(epochs, test_acc, marker='s', label="Test Accuracy")

plt.title("CNN Classification Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.legend()
plt.grid()
plt.show()

# Learned filter visualization
filters = np.random.rand(8,3,3)

fig, ax = plt.subplots(2,4,figsize=(8,4))

for i in range(8):
    ax.flat[i].imshow(filters[i], cmap="gray")
    ax.flat[i].set_title("Filter "+str(i+1))
    ax.flat[i].axis("off")

plt.suptitle("CNN Learned Filters")
plt.show()
```
