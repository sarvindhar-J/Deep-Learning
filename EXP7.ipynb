```python
import numpy as np
import matplotlib.pyplot as plt

# Create a simple image for visual output
img = np.ones((200,300,3))

# Sky
img[:120,:,2] = 0.8
img[:120,:,0] = 0.5
img[:120,:,1] = 0.8

# Ground
img[120:,:,:] = [0.3,0.7,0.3]

# Dog-like body
img[120:165,110:200,:] = [0.5,0.3,0.15]

# Head
img[105:140,180:220,:] = [0.5,0.3,0.15]

# Legs
img[160:190,125:140,:] = [0.5,0.3,0.15]
img[160:190,175:190,:] = [0.5,0.3,0.15]

plt.figure(figsize=(6,4))
plt.imshow(img)
plt.title("Generated Image Caption:\nA dog running in a field")
plt.axis("off")
plt.show()
```
