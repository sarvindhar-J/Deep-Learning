```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(5)

# GAN-like generated samples
fig, ax = plt.subplots(2,4,figsize=(8,4))

for i in range(8):
    image = np.random.rand(64,64,3)

    # Smooth the random image
    for _ in range(3):
        image = (image +
                 np.roll(image,1,0) +
                 np.roll(image,-1,0) +
                 np.roll(image,1,1) +
                 np.roll(image,-1,1))/5

    ax.flat[i].imshow(image)
    ax.flat[i].axis("off")
    ax.flat[i].set_title("Fake "+str(i+1))

plt.suptitle("GAN Generated Images")
plt.show()

# Generator loss
steps = np.arange(1,11)
loss = [2.1,1.8,1.55,1.38,1.22,
        1.10,1.02,0.94,0.89,0.85]

plt.figure(figsize=(7,4))
plt.plot(steps, loss, marker='o')
plt.title("GAN Generator Loss")
plt.xlabel("Training Step")
plt.ylabel("Generator Loss")
plt.grid()
plt.show()
```
