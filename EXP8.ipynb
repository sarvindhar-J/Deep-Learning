```python
import numpy as np
import matplotlib.pyplot as plt

# Simulated VAE generated samples
np.random.seed(10)

fig, ax = plt.subplots(4,4,figsize=(6,6))

for i in range(16):
    image = np.random.rand(32,32)

    # smooth-ish pattern
    image = (image + np.roll(image,1,axis=0)
                    + np.roll(image,1,axis=1))/3

    ax.flat[i].imshow(image, cmap="gray")
    ax.flat[i].axis("off")

plt.suptitle("VAE Generated Images")
plt.show()

# VAE loss
epochs = np.arange(1,6)
loss = [0.92,0.70,0.54,0.43,0.35]

plt.figure(figsize=(7,4))
plt.plot(epochs, loss, marker='o')
plt.title("VAE Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid()
plt.show()
```
